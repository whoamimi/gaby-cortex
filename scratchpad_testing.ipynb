## 3. Mock with Patching

```python
def _create_llm_response(inputs: list):
    responses = []
    for text in inputs:
        responses.append(types.GenerateContentResponse(
            candidates=[
                types.Candidate(
                    content=types.Content(parts=[types.Part(text=text)])
                )
            ]
        ))
    return responses
```

```python
import pytest
from unittest.mock import patch, MagicMock

@patch('app.src.agent.chains.GoogleGenAIProvider', new_callable=MagicMock)
def test_session_controller_llm_class(mock_llm):
    mock_llm.return_value = _create_llm_response(["Testing Response"])[0]
    ss = SessionController(dataset=df)
    print(type(ss.skeleton))
    
test_session_controller_llm_class()
```
